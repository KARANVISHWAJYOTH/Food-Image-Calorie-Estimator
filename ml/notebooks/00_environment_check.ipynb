# ML Environment Check

Run this notebook before starting the ML pipeline. It reports the PyTorch version and CUDA/GPU details. The overall check passes only when PyTorch imports successfully and a CUDA GPU is available. In Colab, select **Runtime > Change runtime type > T4 GPU** (or another GPU) if the check fails for lack of CUDA.

```python
try:
    import torch
except Exception as exc:
    print('PyTorch import: FAIL')
    print(f'  {type(exc).__name__}: {exc}')
    print('Overall: FAIL')
    print('Install the packages from ml/requirements.txt, then restart the runtime.')
else:
    print(f'PyTorch version: {torch.__version__}')
    print(f'PyTorch CUDA runtime: {torch.version.cuda or "not available"}')
    cuda_available = torch.cuda.is_available()
    print(f'CUDA available: {"Yes" if cuda_available else "No"}')

    if cuda_available:
        device_index = torch.cuda.current_device()
        device = torch.cuda.get_device_properties(device_index)
        memory_gib = device.total_memory / (1024 ** 3)
        print(f'GPU: {device.name}')
        print(f'GPU memory: {memory_gib:.2f} GiB')
        print('Overall: PASS')
    else:
        print('GPU: unavailable')
        print('GPU memory: unavailable')
        print('Overall: FAIL')
        print('Enable a GPU runtime, then rerun this cell.')
```
